```python
# 1. IMPORTAÇÃO DAS FERRAMENTAS
import numpy as np
from sklearn.linear_model import LinearRegression
import joblib  # Esta é a ferramenta mágica que salva o modelo
import os
# ---------------------------------------------------------
# FASE 1: TREINAMENTO (O Laboratório)
# ---------------------------------------------------------
# Criando os dados (O "Livro de Estudos")
# X = Entrada (Celsius), y = Saída (Fahrenheit)
X_treino = np.array([[-40], [-10], [0], [8], [15], [22], [38]])
y_treino = np.array([[-40], [14], [32], [46], [59], [72], [100]])
print("🤖 1. Iniciando o treinamento...")
modelo = LinearRegression()
modelo.fit(X_treino, y_treino) # A mágica acontece aqui
print("✅ Modelo treinado com sucesso!")
# ---------------------------------------------------------
# FASE 2: SERIALIZAÇÃO (O "Salvar Como")
# ---------------------------------------------------------
# Agora vamos salvar a "inteligência" num arquivo físico
nome_arquivo = "conversor_temperatura_v1.pkl"
joblib.dump(modelo, nome_arquivo)
print(f"\n💾 2. Modelo salvo no arquivo: {nome_arquivo}")
print(f"   Verifique na aba de arquivos do Colab (ícone da pasta à esquerda).")
print(f"   O arquivo '{nome_arquivo}' apareceu lá? Esse arquivo É o modelo!")
# ---------------------------------------------------------
# FASE 3: INFERÊNCIA (O Uso em Produção)
# ---------------------------------------------------------
# Para provar que funciona, vamos deletar o modelo da memória
del modelo
print("\n🗑️ 3. Modelo deletado da memória RAM.")
# Agora vamos carregar do arquivo (Simulando que estamos em outro computador)
print("📂 4. Carregando o modelo a partir do arquivo...")
modelo_em_producao = joblib.load(nome_arquivo)
# Vamos testar? Quanto é 100°C em Fahrenheit?
temperatura_teste = [[100]]
previsao = modelo_em_producao.predict(temperatura_teste)
print(f"\n🔮 PREVISÃO: 100°C correspondem a aproximadamente {previsao[0][0]:.2f}°F")
print("(A resposta correta seria 212°F. O modelo chegou bem perto!)")
```
